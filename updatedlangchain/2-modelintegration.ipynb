# Streaming and Batch
# Streaming and Batch in LangChain

Imagine you are chatting with an AI assistant. There are two ways the AI can give you answers: streaming and batching.

## 1. Streaming — receiving an answer bit by bit


Think of streaming like watching someone type a message to you. Instead of waiting for the entire answer, you begin seeing the response as it is generated.

Example: You ask an AI, "Explain artificial intelligence."

Instead of waiting for the full explanation, you see:

- AI is...
- AI is a technology...
- AI is a technology that allows computers...

The answer appears gradually until it is complete.

In LangChain, you can use `.stream()` instead of `.invoke()`.

```
from langchain.chat_models import init_chat_modelmodel = init_chat_model("groq:openai/gpt-oss-120b")for chunk in model.stream("Explain artificial intelligence"):    print(chunk.content, end="", flush=True)
```

When is streaming useful?

- Building chatbots that show answers immediately.
- Creating AI applications where users shouldn't wait for the whole response.
- Displaying long answers as they are generated.

## 2. Batch — sending multiple requests together



Think of batching like giving a teacher five exercise books to mark together instead of handing them over one at a time.

You send several questions to the AI in one batch, and LangChain returns a corresponding result for each question.

Example: You want the AI to answer three questions:

1. What is AI?
2. What is machine learning?
3. What is deep learning?

Instead of calling `.invoke()` separately for every question, you can use `.batch()`.

```
questions = [    "What is artificial intelligence?",    "What is machine learning?",    "What is deep learning?"]responses = model.batch(questions)for response in responses:    print(response.content)    print("---")
```

When is batching useful?

- Processing many text documents.
- Classifying many customer reviews.
- Generating summaries for multiple articles.
- Running the same task on several inputs.

## 3. The difference at a glance

| Feature          | Streaming                        | Batch                                 |
| ---------------- | -------------------------------- | ------------------------------------- |
| What it does     | Gives you one response gradually | Processes multiple inputs             |
| Main benefit     | You see the answer sooner        | Convenient for handling many requests |
| LangChain method | `.stream()`                      | `.batch()`                            |
| Example          | Chatbot responding live          | Summarizing 100 documents             |

## 4. Remember it this way

- Streaming = one answer, appearing bit by bit.
- Batch = many inputs, processed together.

One important detail: batching doesn't guarantee that every request runs simultaneously. LangChain's model integration controls how batch requests are executed, and actual speed depends on the provider and its limits.

Also, streaming and batching are not opposites. You can stream a single response or batch multiple inputs; they solve different problems.

````python
# using streaming
model.stream('write a formal letter for me in 200 words')
````

````python
# Display the result as it is generated
for chunk in model.stream("Write a formal letter for me in 200 words"):
    print(chunk.text, end="", flush=True)
````

````python
# Display the result as it is generated
for chunk in model.stream(
    "Write a 200-word paragraph on artificial intelligence."
):
    print(chunk.text, end="", flush=True)
````

````python
chunks = []

for chunk in model.stream(
    "Write a 200-word paragraph on artificial intelligence."
):
    chunks.append(chunk.text)

print("".join(chunks))
````

````python
for chunk in model.stream(
    "Write a 200-word paragraph on artificial intelligence."
):
    print(chunk.text, end="", flush=True)
print()  # Move to a new line after the response finishes
````

````python
from IPython.display import display, Markdown

chunks = []

for chunk in model.stream(
    "Write a 200-word paragraph on artificial intelligence."
):
    chunks.append(chunk.text)

display(Markdown("".join(chunks)))
````

````python
# Batch
responses = model.batch([
    'how are you?',
    'who is donald trump?',
    'how to make egusi soup' 
])
for response in responses:
    print(response.content)
````

````python
# Batch (adding limit to 5 parrallel calls)
responses = model.batch(
    [
        "What is AI?",
        "What is machine learning?",
        "What is deep learning?"
    ],
    config={"max_concurrency": 2}
)
print(responses)
````
